```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('./course_lead_scoring_2026.csv')
df.head()
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
categorical_fields = ['lead_source', 'industry', 'employment_status', 'location']
numerical_fields = ['annual_income', 'number_of_courses_viewed', 'interaction_count', 'lead_score']
```

```python
df[categorical_fields] = df[categorical_fields].fillna('NA')
df[numerical_fields] = df[numerical_fields].fillna(0.0)

df.isna().sum()
```

```python
# Q1
df.industry.mode()
```

```python
# Q2
corr_matrix = df[numerical_fields].corr()
corr_matrix
```

```python
from sklearn.model_selection import train_test_split

# Split the data
df_exclude_target = df.drop('converted', axis='columns')

df_full_train, df_test = train_test_split(df_exclude_target, test_size = 0.2, random_state=42)
df_train, df_val = train_test_split(df_full_train, test_size = 0.25, random_state=42)

df_full_train = df_full_train.reset_index(drop=True)
df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)

y_full_train, y_test = train_test_split(df.converted, test_size = 0.2, random_state=42)
y_train, y_val = train_test_split(y_full_train, test_size = 0.25, random_state=42)
```

```python
from sklearn.metrics import mutual_info_score
```

```python
# Q3
def mutual_info_score_y(series):
  return round(mutual_info_score(series, y_train), 2)

df_train[categorical_fields].apply(mutual_info_score_y).sort_values(ascending=False)
```

```python
train_dict = df_train.to_dict(orient='records')
train_dict[0]
```

```python
from sklearn.feature_extraction import DictVectorizer
```

```python
dv = DictVectorizer(sparse=False)
dv.fit(df_train.to_dict(orient='records'))
```

```python
def prep_X(df):
  df_dict = df.to_dict(orient='records')
  return dv.transform(df_dict)

X_train = prep_X(df_train)
X_full_train = prep_X(df_full_train)
X_val = prep_X(df_val)
X_test = prep_X(df_test)

X_train[0], dv.feature_names_
```

```python
from sklearn.linear_model import LogisticRegression

# Q4
model = LogisticRegression(solver='liblinear', C=1.0, max_iter=1000, random_state=42)
model.fit(X_train, y_train)
```

```python
y_val_pred = model.predict(X_val)
accuracy = (y_val == y_val_pred).mean()
accuracy
```

```python
dict(zip(dv.feature_names_, model.coef_[0].round(3)))
```

```python
# Q5
model.fit(X_train, y_train)
original_accuracy = (model.predict(X_val) == y_val).mean()
original_accuracy
```

```python
for feat in ['lead_score', 'number_of_courses_viewed', 'interaction_count']:
  feat_index = dv.feature_names_.index(feat)
  X_train_elim = np.delete(X_train, feat_index, axis=1)
  X_val_elim = np.delete(X_val, feat_index, axis=1)

  model.fit(X_train_elim, y_train)
  
  y_pred = model.predict(X_val_elim)
  acc = (y_pred == y_val).mean()
  diff = original_accuracy - acc

  print(feat, acc, diff)
```

```python
# Q6
for c in [0.000001, 0.00001, 0.0001, 0.001]:
  reg_model = LogisticRegression(solver='liblinear', C=c, max_iter=1000, random_state=42)
  reg_model.fit(X_train, y_train)
  y_pred = reg_model.predict(X_val)
  acc = (y_pred == y_val).mean()
  print(c, acc)
```
